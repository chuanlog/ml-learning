```python
# 绘制一个sigmoid曲线
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
# 函数(x->y)
def sigmoid(x: np)->list[float]:
    return 1 / (1 + np.exp(-10*x))
# 算数
x=np.linspace(-1,1,100)
y=sigmoid(x)

# 画图
plt.figure(figsize=(8,4))
plt.plot(x, y, label=r'$\sigma(x)=\frac{1}{1+e^{-10*x}}$')
plt.grid(True, alpha=0.3)
plt.axhline(y=0.5, color='r', linestyle='--', alpha=0.6) # y=0.5虚线
plt.axvline(x=0, color='r', linestyle='--', alpha=0.6) # x=0虚线
plt.xlabel("x")
plt.ylabel("sigmoid(x)")
plt.title("Sigmoid Function")
plt.legend()
plt.show()
```
