```python
# 三数之和
class Solution:
    def threeSum(self, nums: list[int]) -> list[list[int]]:
        # 先排序
        nums.sort()
        res=[]
        # 三指针
        left=0
        while left<len(nums):
            if nums[left]>0:
                break
            # 每一个数的第一个
            if left==0 or nums[left - 1]!=nums[left]:
                target=-nums[left]
                mid=left+1
                right=len(nums)-1
                while mid<right:
                    if nums[mid]+nums[right]==target:
                        res.append([nums[left],nums[mid],nums[right]])
                        mid+=1
                        right-=1
                        while mid<right and nums[mid-1]==nums[mid]:
                            mid+=1
                        while mid<right and nums[right+1]==nums[right]:
                            right-=1
                    elif nums[mid]+nums[right]<target:
                        mid+=1
                        while mid<right and nums[mid+1]==nums[mid]:
                            mid+=1
                    elif nums[mid]+nums[right]>target:
                        right-=1
                        while mid<right and nums[right-1]==nums[right]:
                            right-=1
            left+=1
        return res

if __name__ == '__main__':
    s = Solution()
    print(s.threeSum([-1,0,1,2,-1,-4]))
```
